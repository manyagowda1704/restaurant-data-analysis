# Task 4 - Online Delivery

#### Objective
Find the percentage of restaurants that offer online delivery 
and compare the average ratings of restaurants with and 
without online delivery.

Step 1 → Look at the column

Step 2 → Count how many said Yes and No

Step 3 → Calculate percentage of Yes and No

Step 4 → Find average rating for Yes group and No group

Step 5 → Make a chart to compare both groups

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('Restaurant Analysis.xlsx')
df.columns = df.columns.str.replace('ï»¿', '')
```

```python
print(df['Has Online delivery'].value_counts())
```

#### Observation

Out of 9,551 restaurants, only 2,451 offer online delivery 
while 7,100 do not. This means most restaurants in the 
dataset do not have online delivery service.

### Calculating Percentage of Online Delivery

```python
# Total number of restaurants
total = len(df)

# Count Yes and No in delivery column
delivery_counts = df['Has Online delivery'].value_counts()

# Loop through each option and calculate percentage
for option, count in delivery_counts.items():
    percentage = (count / total) * 100
    print(option, "-", round(percentage, 2), "%")
```

#### Observation

74.34% of restaurants do not offer online delivery while 
only 25.66% offer online delivery. This means online delivery is not yet 
widely adopted among restaurants in this dataset.

### Comparing Average Ratings

We compare the average rating of restaurants that offer 
online delivery versus those that do not.

```python
# Split restaurants into Yes/No groups and calculate average rating
avg_rating_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()
print(avg_rating_delivery)
```

#### Observation

Restaurants that offer online delivery have an average rating 
of 3.25 compared to 2.47 for restaurants without online delivery. 
This shows that restaurants with online delivery tend to receive 
higher ratings from customers.

### Visualizing Average Ratings Comparison

```python
avg_rating_delivery.plot(kind='bar', color=['red', 'green'])
plt.title('Average Rating - Online Delivery vs No Delivery')
plt.xlabel('Has Online Delivery')
plt.ylabel('Average Rating')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

#### Observation

The bar chart clearly shows that restaurants offering online 
delivery (3.25) have a higher average rating compared to 
restaurants without online delivery (2.47). The difference 
of 0.78 points is clearly visible in the chart.
