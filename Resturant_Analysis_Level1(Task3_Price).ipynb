# Task 3 - Price Range Distribution

#### Objective
Visualize the distribution of price ranges among restaurants 
and calculate the percentage of restaurants in each price range.

Step 1 → Load data and peek at Price range column

Step 2 → Count restaurants in each price range

Step 3 → Calculate percentage of each price range

Step 4 → Visualize with bar chart

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('Restaurant Analysis.xlsx')
df.columns = df.columns.str.replace('ï»¿', '')
```

```python
print(df['Price range'].value_counts())
```

#### Observation

The price range column has 4 categories from 1 to 4.
Price range 1 (Budget) has the highest count with 4,444 restaurants.
Price range 4 (Luxury) has the lowest count with only 586 restaurants.
This tells us that most restaurants in the dataset are budget friendly.

#### Calculating Percentage of Each Price Range

We calculate what percentage of restaurants fall in each 
price range category.

```python
# Total number of restaurants
total = len(df)

# Count restaurants per price range, sorted 1 to 4
price_counts = df['Price range'].value_counts().sort_index()

# Loop through each price range and calculate percentage
for price, count in price_counts.items():
    percentage = (count / total) * 100
    print(price, "-", round(percentage, 2), "%")
```

#### Observation
Almost half of all restaurants (46.53%) are budget friendly. 
As the price range increases, the number of restaurants decreases. 
Luxury restaurants make up only 6.14% of the dataset.

### Visualizing Price Range Distribution

A bar chart to visually show how restaurants are distributed 
across the four price range categories.

```python
labels = ['Budget(1)', 'Mid-range(2)', 'Premium(3)', 'Luxury(4)']

price_counts.plot(kind='bar', color=['blue', 'orange', 'green', 'red'])
plt.title('Price Range Distribution')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')
# Replace default numbers with readable labels, keep them horizontal
plt.xticks(ticks=[0,1,2,3], labels=labels, rotation=0)
plt.tight_layout()
plt.show()
```

#### Observation

The bar chart shows that 46.53% of restaurants fall in the budget 
category (price range 1), making it the most common price range. 
Mid-range restaurants account for 32.59%, followed by premium at 
14.74% and luxury at only 6.14%. The data shows a clear declining 
pattern — as price range increases, the number of restaurants decreases.
